```python
DATA_PATH = "/kaggle/input/datasets/yelp-dataset/yelp-dataset"

BUSINESS_PATH = f"{DATA_PATH}/yelp_academic_dataset_business.json"
REVIEW_PATH = f"{DATA_PATH}/yelp_academic_dataset_review.json"
CHECKIN_PATH = f"{DATA_PATH}/yelp_academic_dataset_checkin.json"

print(BUSINESS_PATH)
print(REVIEW_PATH)
print(CHECKIN_PATH)
```

```python
#load business.json
import pandas as pd
import numpy as np

business = pd.read_json(
    BUSINESS_PATH,
    lines=True
)

print("Shape:", business.shape)
print("\nColumns:")
print(business.columns.tolist())
```

```python
print("Missing categories:", business["categories"].isna().sum())
```

```python
#Find the cities with the most restaurants
restaurant_business = business[
    business["categories"].fillna("").str.contains(
        "Restaurants",
        case=False,
        regex=False
    )
].copy()

print("Restaurant businesses:", len(restaurant_business))
```

```python
restaurant_city_counts = (
    restaurant_business["city"]
    .value_counts()
    .reset_index()
)

restaurant_city_counts.columns = [
    "city",
    "restaurant_count"
]

print(
    restaurant_city_counts
    .head(30)
    .to_string(index=False)
)
```

```python
#check filtered business count
SELECTED_CITIES = ["Philadelphia", "Tampa"]

yelp_business_sample = business[
    business["categories"].fillna("").str.contains(
        "Restaurants",
        case=False,
        regex=False
    )
    & business["city"].isin(SELECTED_CITIES)
].copy()

print("Filtered business count:", len(yelp_business_sample))
print("\nBusinesses by city:")
print(yelp_business_sample["city"].value_counts())
```

```python
#save the business sample
business_output = "/kaggle/working/yelp_business_sample.csv"

yelp_business_sample.to_csv(
    business_output,
    index=False
)

print("Saved:", business_output)
```

```python
#create the business_id set
business_ids = set(
    yelp_business_sample["business_id"]
)

print("Business rows:", len(yelp_business_sample))
print("Unique business IDs:", len(business_ids))
```

```python
#verify there are no duplicates
print(
    "Duplicate business IDs:",
    yelp_business_sample["business_id"].duplicated().sum()
)
```

```python
CHUNK_SIZE = 100_000

review_chunks = []

total_rows_processed = 0
total_reviews_matched = 0
```

```python
for i, chunk in enumerate(
    pd.read_json(
        REVIEW_PATH,
        lines=True,
        chunksize=CHUNK_SIZE
    )
):
    
    total_rows_processed += len(chunk)

    matched = chunk[
        chunk["business_id"].isin(business_ids)
    ]

    if not matched.empty:
        review_chunks.append(matched)
        total_reviews_matched += len(matched)

    if (i + 1) % 10 == 0:
        print(
            f"Chunks processed: {i + 1:,} | "
            f"Rows processed: {total_rows_processed:,} | "
            f"Matched reviews: {total_reviews_matched:,}"
        )
```

```python
yelp_review_sample = pd.concat(
    review_chunks,
    ignore_index=True
)

print("Review sample shape:", yelp_review_sample.shape)
print("Review count:", len(yelp_review_sample))
```

```python
print("Businesses represented in reviews:",
      yelp_review_sample["business_id"].nunique())

review_business_ids = set(
    yelp_review_sample["business_id"]
)

orphan_review_ids = review_business_ids - business_ids

print("Orphan review business IDs:",
      len(orphan_review_ids))
```

```python
review_output = "/kaggle/working/yelp_review_sample.parquet"

yelp_review_sample.to_parquet(
    review_output,
    index=False,
    compression="snappy"
)

print("Saved:", review_output)
```

## Review Sample Download Preparation

The filtered review sample contains 990,521 reviews and is stored as a single in-memory DataFrame. Because the resulting Parquet file is too large for reliable direct download from the Kaggle interface, it is partitioned into smaller Parquet files for transfer to the local environment.

The partitions are temporary transfer files and will be recombined locally into a single `yelp_review_sample.parquet` dataset.

```python
import os
import glob

output_dir = "/kaggle/working/yelp_review_parts"
os.makedirs(output_dir, exist_ok=True)

# Remove previously generated parts
for f in glob.glob(f"{output_dir}/*.parquet"):
    os.remove(f)

CHUNK_ROWS = 100_000

for i, start in enumerate(
    range(0, len(yelp_review_sample), CHUNK_ROWS),
    start=1
):
    end = min(start + CHUNK_ROWS, len(yelp_review_sample))

    part = yelp_review_sample.iloc[start:end]

    output_path = (
        f"{output_dir}/"
        f"yelp_review_sample_part_{i:02d}.parquet"
    )

    part.to_parquet(
        output_path,
        index=False,
        compression="snappy"
    )

    size_mb = os.path.getsize(output_path) / (1024 ** 2)

    print(
        f"Part {i:02d}: "
        f"{len(part):,} rows | "
        f"{size_mb:.2f} MB"
    )
```

```python
parts = sorted(
    glob.glob(
        "/kaggle/working/yelp_review_parts/*.parquet"
    )
)

print("\nTotal parts:", len(parts))

for p in parts:
    print(
        os.path.basename(p),
        "→",
        round(os.path.getsize(p) / (1024 ** 2), 2),
        "MB"
    )
```
