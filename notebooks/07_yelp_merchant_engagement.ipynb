```python
#import libraries
import pandas as pd
import numpy as np
import glob
import os
```

```python
#define the data paths
data_path = "../data/YELP dataset/"

business_path = os.path.join(data_path, "yelp_business_sample.csv")
checkin_path = os.path.join(data_path, "yelp_checkin_sample.csv")
review_pattern = os.path.join(
    data_path,
    "yelp_review_sample_part*.parquet"
)
```

```python
#find all review chunks
review_files = sorted(glob.glob(review_pattern))

print("Number of review files:", len(review_files))

for file in review_files:
    print(os.path.basename(file))
```

```python
#load and combine the review chunks
review_dfs = [
    pd.read_parquet(file)
    for file in review_files
]

reviews = pd.concat(
    review_dfs,
    ignore_index=True
)

print("Total review rows:", len(reviews))
print("Total review columns:", reviews.shape[1])

del review_dfs
```

```python
#load business and checkin data
business = pd.read_csv(business_path)
checkin = pd.read_csv(checkin_path)

print("Business shape:", business.shape)
print("Check-in shape:", checkin.shape)
print("Review shape:", reviews.shape)
```

```python
#inspect the three database
print("Business columns:")
print(business.columns.tolist())

print("\nCheck-in columns:")
print(checkin.columns.tolist())

print("\nReview columns:")
print(reviews.columns.tolist())
```

```python
# Parse review dates
reviews["date"] = pd.to_datetime(
    reviews["date"],
    errors="coerce"
)

print("Review date range:")
print(reviews["date"].min(), "to", reviews["date"].max())

print("Missing review dates:", reviews["date"].isna().sum())
```

```python
# Make sure missing check-in values are treated as empty strings
checkin["date"] = checkin["date"].fillna("")

# Split comma-separated timestamps into individual rows
checkin = checkin.assign(
    date=checkin["date"].str.split(", ")
).explode("date")

# Convert each check-in timestamp to datetime
checkin["date"] = pd.to_datetime(
    checkin["date"],
    errors="coerce"
)

# Remove rows where the timestamp could not be parsed
checkin = checkin.dropna(subset=["date"])

print("Total check-in events:", len(checkin))
print("Check-in date range:")
print(checkin["date"].min(), "to", checkin["date"].max())

print("Missing check-in dates:", checkin["date"].isna().sum())
```

```python
# Check that business_id exists in all three datasets
print("Business ID in business:", "business_id" in business.columns)
print("Business ID in reviews:", "business_id" in reviews.columns)
print("Business ID in check-ins:", "business_id" in checkin.columns)

# Check for missing business IDs
print("\nMissing business IDs:")
print("Business:", business["business_id"].isna().sum())
print("Reviews:", reviews["business_id"].isna().sum())
print("Check-ins:", checkin["business_id"].isna().sum())
```

```python
# Get the set of valid business IDs
business_ids = set(business["business_id"])

# Check whether reviews/check-ins contain IDs not present in business
review_orphans = (~reviews["business_id"].isin(business_ids)).sum()
checkin_orphans = (~checkin["business_id"].isin(business_ids)).sum()

print("Review orphan rows:", review_orphans)
print("Check-in orphan rows:", checkin_orphans)
```

```python
#check the review date range
review_min_date = reviews["date"].min()
review_max_date = reviews["date"].max()

print("Minimum review date:", review_min_date)
print("Maximum review date:", review_max_date)
```

```python
#define the recent and earlier period
recent_start = review_max_date - pd.DateOffset(months=6)
earlier_start = review_max_date - pd.DateOffset(months=12)

print("Earlier period:", earlier_start, "to", recent_start)
print("Recent period:", recent_start, "to", review_max_date)
```

```python
#align review periods
reviews["period"] = np.select(
    [
        reviews["date"] >= recent_start,
        (reviews["date"] >= earlier_start) & (reviews["date"] < recent_start)
    ],
    [
        "recent",
        "earlier"
    ],
    default="outside"
)

print(reviews["period"].value_counts())
```

```python
#align checkin periods
checkin["period"] = np.select(
    [
        checkin["date"] >= recent_start,
        (checkin["date"] >= earlier_start) & (checkin["date"] < recent_start)
    ],
    [
        "recent",
        "earlier"
    ],
    default="outside"
)

print(checkin["period"].value_counts())
```

### review metrics:

```python
#total reviews
review_total = (
    reviews
    .groupby("business_id")
    .size()
    .rename("total_reviews")
    .reset_index()
)

review_total.head()
```

```python
#monthly review count
reviews["review_month"] = reviews["date"].dt.to_period("M")
monthly_reviews = (
    reviews
    .groupby(["business_id", "review_month"])
    .size()
    .rename("monthly_review_count")
    .reset_index()
)

monthly_reviews.head()
```

```python
#recent vs earlier review volume
period_reviews = (
    reviews[reviews["period"].isin(["recent", "earlier"])]
    .groupby(["business_id", "period"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

if "recent" not in period_reviews.columns:
    period_reviews["recent"] = 0

if "earlier" not in period_reviews.columns:
    period_reviews["earlier"] = 0

period_reviews = period_reviews.rename(
    columns={
        "recent": "recent_reviews",
        "earlier": "earlier_reviews"
    }
)

period_reviews.head()
```

```python
#review growth
period_reviews["review_growth"] = np.where(
    period_reviews["earlier_reviews"] > 0,
    (
        period_reviews["recent_reviews"]
        - period_reviews["earlier_reviews"]
    ) / period_reviews["earlier_reviews"],
    np.nan
)

period_reviews.head()
```

```python
#average rating
average_rating = (
    reviews
    .groupby("business_id")["stars"]
    .mean()
    .rename("average_rating")
    .reset_index()
)

average_rating.head()
```

```python
#monthly average rating
monthly_rating = (
    reviews
    .groupby(["business_id", "review_month"])["stars"]
    .mean()
    .rename("monthly_average_rating")
    .reset_index()
)

monthly_rating.head()
```

```python
#recent vs earlier average rating
period_rating = (
    reviews[reviews["period"].isin(["recent", "earlier"])]
    .groupby(["business_id", "period"])["stars"]
    .mean()
    .unstack()
    .reset_index()
)

if "recent" not in period_rating.columns:
    period_rating["recent"] = np.nan

if "earlier" not in period_rating.columns:
    period_rating["earlier"] = np.nan

period_rating = period_rating.rename(
    columns={
        "recent": "recent_average_rating",
        "earlier": "earlier_average_rating"
    }
)

period_rating.head()
```

```python
#rating change
period_rating["rating_change"] = (
    period_rating["recent_average_rating"]
    - period_rating["earlier_average_rating"]
)

period_rating.head()
```

### Checkin Metrics:

```python
#total checkins
checkin_total = (
    checkin
    .groupby("business_id")
    .size()
    .rename("total_checkins")
    .reset_index()
)

checkin_total.head()
```

```python
#monthly checkins
checkin["checkin_month"] = checkin["date"].dt.to_period("M")
monthly_checkins = (
    checkin
    .groupby(["business_id", "checkin_month"])
    .size()
    .rename("monthly_checkin_count")
    .reset_index()
)

monthly_checkins.head()
```

```python
#recent vs earlier checkins
period_checkins = (
    checkin[checkin["period"].isin(["recent", "earlier"])]
    .groupby(["business_id", "period"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

if "recent" not in period_checkins.columns:
    period_checkins["recent"] = 0

if "earlier" not in period_checkins.columns:
    period_checkins["earlier"] = 0

period_checkins = period_checkins.rename(
    columns={
        "recent": "recent_checkins",
        "earlier": "earlier_checkins"
    }
)

period_checkins.head()
```

```python
#checkin growth
period_checkins["checkin_growth"] = np.where(
    period_checkins["earlier_checkins"] > 0,
    (
        period_checkins["recent_checkins"]
        - period_checkins["earlier_checkins"]
    ) / period_checkins["earlier_checkins"],
    np.nan
)

period_checkins.head()
```

### Assemble the merchant engagement dataset

```python
#start with the business table
merchant_engagement = business.copy()

print("Starting rows:", len(merchant_engagement))
```

```python
#merge review metrics
merchant_engagement = merchant_engagement.merge(
    review_total,
    on="business_id",
    how="left"
)

merchant_engagement = merchant_engagement.merge(
    period_reviews,
    on="business_id",
    how="left"
)

merchant_engagement = merchant_engagement.merge(
    average_rating,
    on="business_id",
    how="left"
)

merchant_engagement = merchant_engagement.merge(
    period_rating,
    on="business_id",
    how="left"
)
```

```python
#merge checkin metrics
merchant_engagement = merchant_engagement.merge(
    checkin_total,
    on="business_id",
    how="left"
)

merchant_engagement = merchant_engagement.merge(
    period_checkins,
    on="business_id",
    how="left"
)
```

```python
#check the resulting shape
print("Merchant engagement shape:", merchant_engagement.shape)
print("Number of businesses:", merchant_engagement["business_id"].nunique())
```

```python
merchant_engagement.head()
```

```python
#Fill missing review/check-in counts with zero not NaN
count_columns = [
    "total_reviews",
    "recent_reviews",
    "earlier_reviews",
    "total_checkins",
    "recent_checkins",
    "earlier_checkins"
]

merchant_engagement[count_columns] = (
    merchant_engagement[count_columns]
    .fillna(0)
    .astype(int)
)
```
