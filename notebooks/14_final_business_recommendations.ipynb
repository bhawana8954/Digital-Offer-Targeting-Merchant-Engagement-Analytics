## Criteo Customer Targeting — Results Reference

The Criteo analysis provides three complementary findings: advertising exposure generates incremental conversions, certain customer segments respond more strongly, and incremental response is concentrated among a smaller group of high-uplift customers.

### Overall Incrementality

| Metric                    |        Result |
| ------------------------- | ------------: |
| Control conversion rate   |       0.1938% |
| Treatment conversion rate |       0.3089% |
| Absolute lift             |     0.1152 pp |
| Relative lift             |        59.45% |
| P-value                   | 7.31 × 10⁻¹⁷⁹ |

Treatment exposure produced a statistically significant positive incremental effect, with a 59.45% relative improvement in conversion over control.

### High-Response Segments

Four feature-defined segments were classified as attractive:

| Feature | Segment           | Relative Lift |
| ------- | ----------------- | ------------: |
| f8      | (3.634, 3.911]    |        54.23% |
| f6      | (-31.431, -6.699] |        70.28% |
| f0      | (12.615, 21.923]  |        62.14% |
| f2      | (8.213, 8.723]    |        58.49% |

The **f8 segment** had the strongest absolute lift at 0.4043 percentage points, while **f6** had the highest relative lift at 70.28%.

### Customer-Level Targeting

A T-learner was used to estimate customer-level incremental response and rank customers by predicted uplift. The **Top 10%** of customers achieved an observed lift of **0.005672 (0.57 percentage points)** and captured approximately **65% of observed conversions**.

| Targeting Group | Population Share | Observed Lift |
| --------------- | ---------------: | ------------: |
| Top 10%         |              10% |       0.57 pp |
| Top 20%         |              20% |       0.28 pp |
| Top 30%         |              30% |       0.12 pp |
| Bottom 50%      |              50% |       0.15 pp |

The Top 10% showed approximately **3.8× higher observed uplift than the Bottom 50%**. Because uplift was not perfectly monotonic across all groups, the results support prioritizing the highest-uplift customers while continuing to validate targeting performance through controlled experimentation.

### Recommendation Evidence

Overall, the analysis supports three conclusions:

1. **Exposure works:** treatment generated a statistically significant incremental effect.
2. **Response varies across segments:** four feature-defined segments showed particularly strong performance.
3. **Incremental response is concentrated:** the highest-uplift customers produced the strongest observed targeting results.

## 20.2 Merchant Recommendations

### Merchant Health Across the Portfolio

The full merchant portfolio contains **8,812 merchants**: 4,266 Stable (48.41%), 2,185 Growing (24.80%), and 2,361 Declining (26.79%). More than one-quarter of merchants show declining engagement and may require additional attention.

### Priority-Based Recommendations

The 300-merchant sentiment sample combines engagement and customer sentiment to identify four priority groups:

| Priority Group      | Merchants |  Share |
| ------------------- | --------: | -----: |
| Expand              |       147 | 49.00% |
| Growth Opportunity  |        79 | 26.33% |
| Monitor / Intervene |        53 | 17.67% |
| Reassess            |        21 |  7.00% |

### Recommended Actions

* **Monitor / Intervene:** Focus on service quality and customer experience. These merchants have relatively strong engagement but negative sentiment, so the priority is to protect existing engagement while addressing recurring CX issues.

* **Growth Opportunity:** Focus on increasing engagement through targeted promotions, loyalty initiatives, and campaigns. Positive sentiment provides a favorable foundation for increasing customer activity.

* **Expand:** Continue investment where both engagement and sentiment support further growth.

* **Reassess:** Review merchant performance and customer feedback before allocating significant additional resources.

Overall, the recommendations distinguish between **protecting existing engagement through CX improvement** and **building engagement where customer sentiment is already favorable**, allowing merchant resources to be allocated according to the underlying risk and opportunity.
